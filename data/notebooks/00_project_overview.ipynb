## Future17 AI Governance Risk Toolkit for UK Higher Education

This notebook explains the project scope, repo logic, datasets, and workflow.

The project is **not** a student prediction model. It is a governance analytics toolkit for auditing AI systems used in higher education.

Core use cases:

- AI detection / academic integrity tools
- Student-facing chatbots
- VLE engagement analytics
- Career-service AI tools
- AI-assisted marking and feedback pilots
- Early-alert student support systems

## Core Governance Question

> Are universities deploying AI systems in ways that unintentionally disadvantage students because of unequal AI access, AI confidence, or AI proficiency?

The toolkit demonstrates how a university could assess AI systems through:

```text
AI system inventory
→ data / logs / evidence
→ representation analysis
→ fairness metrics
→ proxy bias detection
→ governance evidence validation
→ risk scoring
→ deployment recommendation
```
